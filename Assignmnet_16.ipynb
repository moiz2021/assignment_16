**Q1. Write a query to list the top 10 genres with the highest number of tracks. (Show in bar plot)**

```python
query='''
select g.name genre_name,count(t.track_id) track_count
from genre g
inner join track t
on g.genre_id = t.genre_id
group by g.genre_id, g.name
order by track_count DESC
limit 10
'''
df=sql_query(query)
plt.figure(figsize=(10, 6))
sns.barplot(
    data=df, 
    x='track_count', 
    y='genre_name', 
    palette='viridis'
)

plt.title('Top 10 Genres by Number of Tracks', fontsize=14, fontweight='bold')
plt.xlabel('Number of Tracks', fontsize=12)
plt.ylabel('Genre', fontsize=12)
plt.tight_layout()

plt.show()
```

**Q2. Write a query to list the top 10 albums that contain the most tracks**

```python
query='''
select a.title album_title, count(t.track_id) total_tracks
from album a
inner join track t
on a.album_id = t.album_id
group by a.album_id, a.title
order by total_tracks desc
limit 10
'''
sql_query(query)
```

**Q3. Write a query to find the top 10 customers who spent the most money.**

```python
query='''
select c.first_name || ' ' || c.last_name customer_name, sum(i.total) total_spent
from customer c
inner join invoice i
on c.customer_id = i.customer_id
group by c.customer_id, customer_name
order by total_spent desc
limit 10
'''
sql_query(query)
```

**Q4. Write a query to find the top 5 artists who generated the highest revenue from track sales.**

```python
query='''
select ar.name artist_name, sum(il.unit_price * il.quantity) total_revenue
from artist ar
inner join album al
on ar.artist_id = al.artist_id
inner join track t
on al.album_id = t.album_id
inner join invoice_line il
on t.track_id = il.track_id
group by ar.artist_id, ar.name
order by total_revenue desc
limit 5
'''
sql_query(query)
```

**Q5. Countries by Average Revenue per Invoice**

```python
query='''
select billing_country country, avg(total) avg_revenue
from invoice
group by billing_country
order by avg_revenue desc
'''
sql_query(query)
```

**Q6. Write a query to find the media type with number of tracks.**

```python
query='''
select m.name media_type, count(t.track_id) track_count
from media_type m
inner join track t
on m.media_type_id = t.media_type_id
group by m.media_type_id, m.name
order by track_count desc
'''
sql_query(query)
```

**Q7. Which artists have released tracks in more than one genre?**

```python
query='''
select ar.name artist_name, count(distinct g.genre_id) genre_count
from artist ar
inner join album al
on ar.artist_id = al.artist_id
inner join track t
on al.album_id = t.album_id
inner join genre g
on t.genre_id = g.genre_id
group by ar.artist_id, ar.name
having count(distinct g.genre_id) > 1
order by genre_count desc
'''
sql_query(query)
```

**Q8. Write a query to list the top 10 longest tracks by duration.**

```python
query='''
select t.name track_name, t.milliseconds
from track t
order by t.milliseconds desc
limit 10
'''
sql_query(query)
```

**Q9. Top 5 Revenue Share percentage by Artist (Pie Chart)**

```python
query='''
select ar.name artist_name, sum(il.unit_price * il.quantity) total_revenue
from artist ar
inner join album al on ar.artist_id = al.artist_id
inner join track t on al.album_id = t.album_id
inner join invoice_line il on t.track_id = il.track_id
group by ar.artist_id, ar.name
order by total_revenue desc
limit 5
'''
df=sql_query(query)
plt.figure(figsize=(4, 4))
plt.pie(
    df['total_revenue'], 
    labels=df['artist_name'], 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=plt.cm.tab10.colors[:5]
)

plt.title('Top 5 Artists Revenue Share (%)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Q10. Revenue by Genre (Rock, Latin, Metal) in Each Country (Stacked Column Chart)**

```python
query='''
select 
        i.billing_country country,
        sum(case when g.name = 'Rock' then il.unit_price * il.quantity else 0 end) rock,
        sum(case when g.name = 'Latin' then il.unit_price * il.quantity else 0 end) latin,
        sum(case when g.name = 'Metal' then il.unit_price * il.quantity else 0 end) metal
    from invoice i
    inner join invoice_line il on i.invoice_id = il.invoice_id
    inner join track t on il.track_id = t.track_id
    inner join genre g on t.genre_id = g.genre_id
    where g.name in ('Rock', 'Latin', 'Metal')
    group by i.billing_country
    order by (rock + latin + metal) desc
'''
df=sql_query(query)
df.set_index('country')[['rock', 'latin', 'metal']].plot(
    kind='bar', 
    stacked=True, 
    figsize=(12, 6),
    colormap='viridis'
)

plt.title('Revenue by Genre (Rock, Latin, Metal) per Country', fontsize=14, fontweight='bold')
plt.xlabel('Country', fontsize=12)
plt.ylabel('Revenue ($)', fontsize=12)
plt.legend(title='Genre')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```
